# 1. Exploratory Data Analysis (EDA) & Feature Selection
In this notebook, we load the processed tabular dataset containing our extracted time and frequency domain features. We will analyze the distributions, check for multicollinearity, and select the best features for our Machine Learning models.


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.append(os.path.abspath('..'))
from src import config

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context('notebook')
```

## Load Data


```python
data_path = config.PROCESSED_DATA_DIR / 'features.csv'
if not os.path.exists(data_path):
    print("Features not found! Please run `python src/data_processing.py` first.")
else:
    df = pd.read_csv(data_path)
    print(f"Data shape: {df.shape}")
    display(df.head())
```

## Target Variable Distribution
Let's see how our Chatter Index (CI) maps to the binary 'Label' (0=Stable, 1=Chatter).


```python
if 'df' in locals():
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    sns.histplot(df, x='CI', hue='Label', bins=20, kde=True, ax=axes[0], palette='Set1')
    axes[0].set_title('Chatter Index Distribution')
    
    sns.countplot(data=df, x='Label', ax=axes[1], palette='Set1')
    axes[1].set_title('Class Imbalance Check (0=Stable, 1=Chatter)')
    
    plt.tight_layout()
    plt.show()
```

## Feature Correlation
Checking for multicollinearity amongst our extracted features.


```python
if 'df' in locals():
    plt.figure(figsize=(12, 10))
    corr = df[config.FEATURE_NAMES].corr()
    mask = np.triu(np.ones_like(corr, dtype=bool))
    sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
    plt.title('Feature Correlation Matrix')
    plt.show()
```

## Conclusion
We observe which features are highly correlated (e.g. RMS and Std might be identical if Mean is near 0). We will use all features initially and let tree-based models or regularization handle selection.
